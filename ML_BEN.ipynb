### 1. Inicializar um novo repositório Git local

Se você ainda não tem um repositório Git local, use este comando para inicializar um.

```python
# Navegue até o diretório do seu projeto (se ainda não estiver lá)
# %cd /content/my_project_folder

# Inicializa um novo repositório Git no diretório atual
!git init
```

### 2. Adicionar arquivos ao 'staging area' e fazer o commit inicial

Isso prepara seus arquivos para serem incluídos no histórico do Git e, em seguida, os salva com uma mensagem de commit.

```python
git status
```

```python
# Adiciona todos os arquivos do diretório atual (e subdiretórios) ao staging area
!git add .

# Faz o commit dos arquivos adicionados com uma mensagem
!git commit -m "Primeiro commit"
```

### 3. Adicionar um repositório remoto (ex: GitHub, GitLab)

Você precisará criar um repositório vazio em um serviço como GitHub ou GitLab. Depois, adicione a URL como um 'remote' chamado `origin`.

```python
# Substitua <URL_DO_SEU_REPOSITORIO_REMOTO> pela URL real do seu repositório remoto (ex: https://github.com/seu_usuario/seu_repositorio.git)
!git remote add origin <https://github.com/PacktPublishing/Machine-Learning-for-Time-Series-with-Python.git>

# Opcional: Verifique se o remote foi adicionado corretamente
!git remote -v
```

### 4. Enviar (push) seus commits locais para o repositório remoto

Isso publica seu histórico de commits local para o branch `master` (ou `main`) do repositório remoto `origin`.

```python
# Envia o branch atual (geralmente master ou main) para o remote 'origin'
# O '-u' define 'origin/master' como o upstream, para que 'git push' e 'git pull' funcionem sem argumentos futuros.
!git push -u origin master
# Se o seu branch principal for 'main', use: !git push -u origin main
```

### 5. Puxar (pull) e Enviar (push) atualizações

Após a configuração inicial, você usará `pull` para obter as últimas mudanças do repositório remoto e `push` para enviar suas novas alterações.

```python
# Para puxar as últimas mudanças do repositório remoto para o seu branch atual
!git pull origin master
# Ou se estiver em outro branch: !git pull origin <nome_do_seu_branch>
```

```python
# Para enviar suas novas mudanças (após 'git add' e 'git commit') para o repositório remoto
!git push origin master
# Ou se estiver em outro branch: !git push origin <nome_do_seu_branch>
```

Lembre-se de substituir os placeholders como `<URL_DO_SEU_REPOSITORIO_REMOTO>` e `<nome_do_seu_branch>` pelos valores corretos do seu projeto!
